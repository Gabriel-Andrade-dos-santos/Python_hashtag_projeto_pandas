```python
import pandas as pd
import numpy as np
import datetime

vendas_df = pd.read_csv("vendas_tech.csv", low_memory=False)
gerentes_df = pd.read_excel("gerentes_lojas.xlsx" )

display(vendas_df)
print("-"*200)
display(gerentes_df)
```

```python
#Inspeção
display(vendas_df.head())
print("-"*200)
display(vendas_df.tail())
print("-"*200)
display(vendas_df.sample())
print("-"*200)
display(vendas_df.shape)
print("-"*200)
display(vendas_df.info())
```

```python
#Tratamento de dados

#colunas
vendas_analise = vendas_df.drop(columns=["Data_Base"])


#nulos
#vendas_analise = vendas_analise.dropna() -> exclui todas as linhas que tem um valor vazio
vendas_analise["Loja"] = vendas_analise["Loja"].fillna("Online")

#tipo de dados
vendas_analise["Data"] = pd.to_datetime(vendas_analise["Data"], format="%Y-%m-%d")

#padronização
vendas_analise["Loja"] = vendas_analise["Loja"].str.strip()
vendas_analise["Loja"] = vendas_analise["Loja"].str.title()

gerentes_df["Loja"] = gerentes_df["Loja"].str.strip()
gerentes_df["Loja"] = gerentes_df["Loja"].str.title()

#Duplicatas
vendas_analise = vendas_analise.drop_duplicates(subset=["ID_Pedido"])
display(vendas_analise)
```

```python
#Novas colunas

#Faturamento
vendas_analise["Faturamento"] = vendas_analise["Preco_Unitario"] * vendas_analise["Qtd"]

#Região
display(vendas_analise["Loja"].unique())
dict_regiao = {
    'São Paulo' : 'Sudeste',
    'Belo Horizonte' : 'Sudeste',
    'Online' : 'Online',
    'Rio De Janeiro' : 'Sudeste',
    'Salvador' : 'Nordeste',
    'Recife' : 'Nordeste',
    'Curitiba' : 'Sul',
    'Porto Alegre' : 'Sul'

}

vendas_analise["Região"] = vendas_analise["Loja"].map(dict_regiao)
#Forma de venda
vendas_analise["Forma_de_venda"] = np.where(vendas_analise["Loja"] == "Online", "Online", "Presencial")
display(vendas_analise)
```

```python
#Analizar

vendas_analise = vendas_analise.sort_values(by=["Data"], ascending=False)
vendas_analise = vendas_analise.reset_index(drop=True)
#display(vendas_analise)
#.loc -> por nome da coluna e nome da linha
loja = vendas_analise.loc[3, "Loja"]
produto = vendas_analise.loc[3, "Produto"]
cliente = vendas_analise.loc[3, "Cliente"]
print(loja, produto, cliente)
#.iloc -> por posição
#condicao

#df_id_4 = vendas_analise[vendas_analise["ID_Pedido"] == 4]
#display(df_id_4)
#Exportar pedaços da base

df_vendas_sp = vendas_analise[vendas_analise["Loja"] == "São Paulo"]
df_vendas_sp.to_csv("Vendas_Sao_Paulo.csv",index=False, sep=";")

#exportar as vendas de 2024
df_vendas_2024 = vendas_analise[vendas_analise["Data"].dt.year == 2024]



df_hdmi_sul = vendas_analise[(vendas_analise["Produto"] == "Cabo HDMI") & (vendas_analise["Região"] == "Sul")]

# & == E
# | == OU

display(df_hdmi_sul)
```

```python
rank_vendas = vendas_analise[["Loja", "Faturamento"]].groupby("Loja").sum(numeric_only=True)
rank_vendas = rank_vendas.sort_values(by="Faturamento", ascending=False)
rank_vendas = rank_vendas.reset_index()
rank_vendas["Faturamento"] = rank_vendas["Faturamento"].map("R${:,.2f}".format)
display(rank_vendas)
print()
```

```python
#mais venderam no Online
df_vendas_online = vendas_analise[vendas_analise["Loja"] == "Online"]
rank_vendas_online = df_vendas_online[["Produto", "Faturamento"]].groupby("Produto").sum(numeric_only=True)
rank_vendas_online = rank_vendas_online.sort_values(by="Faturamento", ascending=False)
rank_vendas_online = rank_vendas_online.reset_index()
rank_vendas_online["Faturamento"] = rank_vendas_online["Faturamento"].map("R${:,.2f}".format)
display(rank_vendas_online)
```

```python
#Analize de rank por loja e produto

analise_produtos_em_lojas = vendas_analise[["Loja", "Produto", "Qtd"]].groupby(["Loja", "Produto"]).sum()
analise_lojas_em_prdutos = vendas_analise[["Loja", "Produto", "Qtd"]].groupby(["Produto", "Loja"]).sum()
display(analise_produtos_em_lojas)
display(analise_lojas_em_prdutos)
```

```python
#Quais gerentes bateram a meta Jan 2023?
df_vendas_gerentes = vendas_analise[(vendas_analise["Data"].dt.year==2023) & (vendas_analise["Data"].dt.month == 1)]

df_vendas_gerentes = df_vendas_gerentes[["Loja", "Faturamento"]].groupby("Loja", as_index=False).sum(numeric_only=True)
df_vendas_gerentes = df_vendas_gerentes.merge(gerentes_df,on="Loja", how="left")
df_vendas_gerentes["Bateu_Meta"] = np.where(df_vendas_gerentes["Faturamento"] >= df_vendas_gerentes["Meta_Mensal"], "Sim","Não")
display(df_vendas_gerentes)
```
